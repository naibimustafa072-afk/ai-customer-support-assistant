```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate large-scale simulated German retail data
np.random.seed(2026)
dates = pd.date_range(start="2024-01-01", end="2026-10-01", freq="D")
n_rows = 10000

data = {
    'Date': np.random.choice(dates, n_rows),
    'City': np.random.choice(['München', 'Berlin', 'Frankfurt', 'Hamburg'], n_rows),
    'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Home_Goods', 'Automotive'], n_rows),
    'Units_Sold': np.random.randint(1, 50, n_rows),
    'Price_Euro': np.random.uniform(5.0, 500.0, n_rows)
}

df_large = pd.DataFrame(data)
df_large['Total_Revenue'] = df_large['Units_Sold'] * df_large['Price_Euro']

# 2. Advanced Engineering Analysis: Revenue by City and Category
plt.figure(figsize=(12, 6))
sns.barplot(data=df_large, x='City', y='Total_Revenue', hue='Product_Category', palette='viridis')

plt.title('Advanced Enterprise Dashboard: Total Revenue Analysis across Germany (2024-2026)', fontsize=14, fontweight='bold')
plt.xlabel('German Cities / Branches', fontsize=12)
plt.ylabel('Total Revenue in Euro (€)', fontsize=12)
plt.legend(title='Product Categories', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 3. Show data preview and the plot
print("--- Large Scale Enterprise Data Preview ---")
print(df_large.head())
print(f"\nTotal Dataset Rows: {len(df_large)}")
print("="*50)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Feature Engineering: Convert text columns to numbers for AI processing
le_city = LabelEncoder()
le_cat = LabelEncoder()

df_large['City_Encoded'] = le_city.fit_transform(df_large['City'])
df_large['Category_Encoded'] = le_cat.fit_transform(df_large['Product_Category'])

# Define Inputs (Features) and Output (Target)
X_ai = df_large[['City_Encoded', 'Category_Encoded', 'Price_Euro']]
y_ai = df_large['Units_Sold']

# 2. Split data into Training (80%) and Testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X_ai, y_ai, test_size=0.2, random_state=42)

# 3. Initialize and train the Advanced Random Forest Regressor
ai_model = RandomForestRegressor(n_estimators=100, random_state=42)
ai_model.fit(X_train, y_train)

# 4. Evaluate the Model Performance
y_pred = ai_model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("="*60)
print("🚀 ENTERPRISE MACHINE LEARNING MODEL TRAINED SUCCESSFULLY 🚀")
print(f"📊 Model Accuracy Metric (R2 Score): {r2:.4f}")
print(f"📉 Mean Absolute Error (MAE): {mae:.2f} Units")
print("="*60)

# 5. Live Scenario Optimization: Predict sales for a premium product in München
# Let's test a premium Electronics item (Price: 450€) in München
test_city = le_city.transform(['München'])[0]
test_cat = le_cat.transform(['Electronics'])[0]
test_price = 450.0

optimized_prediction = ai_model.predict([[test_city, test_cat, test_price]])

print(f"\n🔮 OPTIMIZATION INSIGHT FOR GERMAN OPERATIONS:")
print(f"Target Branch: München | Category: Electronics | Price: 450.00 €")
print(f"Recommended Inventory Supply for Next Month: {int(np.ceil(optimized_prediction[0]))} Units")
print("="*60)
```
